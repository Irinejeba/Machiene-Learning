### **DECISION TREE**

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# 1. Load the dataset
# Note: Update the path string if you move your CSV file to a different folder
file_path = r"C:\Users\Admin\Desktop\CS\ML Class\Movie Interest.csv"
move_info = pd.read_csv(file_path)
```

```python
# 2. Explore the dataset (Optional print statements for VS Code terminal)
print("--- Dataset Shape ---")
print(move_info.shape)
```

```python
print("\n--- Missing Values ---")
print(move_info.isna().sum())
```

```python
print("\n--- DataFrame Info ---")
move_info.info()
```

```python
print("\n--- Descriptive Statistics ---")
print(move_info.describe())
```

```python
# 3. Split data into Features (X) and Target (y)
input_dataset = move_info.drop(columns=["Interest"])
output_dataset = move_info["Interest"]
```

```python
# 4. Scenario A: Train model on the entire dataset and make a raw prediction
print("\n--- Running Scenario A: Full Training ---")
move_model_full = DecisionTreeClassifier()
move_model_full.fit(input_dataset, output_dataset)

# Predict for [Age=16, Gender=1] and [Age=10, Gender=0]
predictions_full = move_model_full.predict([[16, 1], [10, 0]])
print("Predictions (Full Model):", predictions_full)
```

```python
# 5. Scenario B: Split data, train, evaluate accuracy, and predict
print("\n--- Running Scenario B: Train/Test Split ---")
# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    input_dataset, output_dataset, test_size=0.2, random_state=42
)

move_model_split = DecisionTreeClassifier()
move_model_split.fit(X_train, y_train)

# Evaluate model performance
test_predictions = move_model_split.predict(X_test)
accuracy = accuracy_score(y_test, test_predictions)
print(f"Model Accuracy: {accuracy * 100:.2f}%")

# Predict using specific values on the split model
predictions_split = move_model_split.predict([[15, 1], [10, 0]])
print("Predictions (Split Model):", predictions_split)
```
