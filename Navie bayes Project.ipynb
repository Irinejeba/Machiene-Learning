```python
# Drop that 1 empty row
spam_data = spam_data.dropna(subset=['Body', 'Label'])

input_data = spam_data['Body'].astype(str)   # NOT 'text', your column is Body
output_data = spam_data['Label'].astype(str)  # NOT 'label' small, your column is Label
```

```python
from sklearn.model_selection import train_test_split
input_data_train, input_data_test, output_data_train, output_data_test = train_test_split(input_data, output_data, test_size=0.3)

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import make_pipeline

model = make_pipeline(TfidfVectorizer(), MultinomialNB())
model.fit(input_data_train, output_data_train)
```

```python
predicted_spam = model.predict(input_data_test)
```

```python
predicted_spam
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(output_data_test, predicted_spam)
```

```python
import pandas as pd
matrix = pd.DataFrame(confusion_matrix(output_data_test, predicted_spam), columns=["predicted_Ham", "predicted_Spam"], index=["Actual_Ham", "Actual_Spam"])
matrix
```

```python
from sklearn.metrics import accuracy_score, classification_report

accuracy_info = accuracy_score(output_data_test, predicted_spam)
print("Accuracy:", accuracy_info)
print(classification_report(output_data_test, predicted_spam))
```
